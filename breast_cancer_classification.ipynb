# **11. CHECK THE CORRELATION BETWEEN THE NUMERIC COLUMNS**

```python
plt.figure(figsize=(35,35))
sns.heatmap(df.corr(numeric_only=True,method="pearson"),annot=True)
```

CONCLUSION : THERE ARE MANY COLUMNS THAT SHOW HIGH CORRELATION WITH OTHER COLUMN SO MY APPROACH IS TO SELECT ONE COLUMN AND REMOVE OTHER COLUMNS IF THEY ARE HIGHLY CORRELATED WITH EACH OTHER.(Criteria if correlation> 0.8)

```python
corr_cols=df.select_dtypes(include="number").columns
drop_cols=[]
for i,col in enumerate(corr_cols):
  if col in drop_cols:
    continue
  for j,other_cols in enumerate(corr_cols[i+1:],start=i+1):
    if other_cols in drop_cols:
      continue
    if df[col].corr(df[other_cols])>0.8:
        drop_cols.append(other_cols)
print(drop_cols)
```

```python
print(f"Total columns that I want to drop based on above criteria {len(drop_cols)}")
```

# **13. TRAIN TEST SPLIT**

```python
X=data.drop(columns="diagnosis")
y=data["diagnosis"].map({"B": 0,"M":1})
# Mapping is important here because xgboost and lightgbm does not work in categorical column
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42,stratify=y)
```

# **14. PREPROCESSING OF DATA AND BUILDING PIPELINES**

```python
from sklearn.preprocessing import StandardScaler,FunctionTransformer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
numeric_cols=['radius_mean', 'texture_mean', 'smoothness_mean', 'compactness_mean',
       'symmetry_mean', 'fractal_dimension_mean', 'radius_se', 'texture_se',
       'smoothness_se', 'compactness_se', 'concave points_se', 'symmetry_se',
       'symmetry_worst', 'fractal_dimension_worst']
```

```python
numeric_pipeline=Pipeline(steps=[("log",FunctionTransformer(np.log1p)),
                                 ("scaler",StandardScaler())])
```

# **14. CREATE COLUMN TRANSFORMER**

```python
preprocessor=ColumnTransformer(transformers=[("numeric",numeric_pipeline,numeric_cols)])
preprocessor
```

# **15. IMPORT LIBRARIES FOR MODEL TRAINING**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVC
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
```

# **16. MODEL TRAINING AND EVALUATION**

```python
models={
    'LogisticRegression':LogisticRegression(class_weight='balanced',random_state=42),
    'DecisionTree':DecisionTreeClassifier(class_weight='balanced',random_state=42),
    'RandomForest':RandomForestClassifier(class_weight='balanced',random_state=42),
    'GradientBoosting':GradientBoostingClassifier(random_state=42),
    'SVC':SVC(class_weight='balanced',probability=True,random_state=42),
    'XGBoost':XGBClassifier(n_estimators=500,learning_rate=0.03,random_state=42),
    'LGBM':LGBMClassifier(random_state=42,learning_rate=0.05,n_estimators=5000,verbose=-1)
}

for name,model in models.items():
   pipe=Pipeline(steps=[
       ('preprocessor',preprocessor),('classifier',model)
   ])
   accuracy=cross_val_score(pipe,X_train,y_train,cv=5,scoring='accuracy')
   macro_f1=cross_val_score(pipe,X_train,y_train,cv=5,scoring='f1_macro')
   recall_score=cross_val_score(pipe,X_train,y_train,cv=5,scoring='recall_macro')
   precision_score=cross_val_score(pipe,X_train,y_train,cv=5,scoring='precision_macro')
   print(f"*"*25,name,"*"*25)
   print(f"{name} --> Accuracy {accuracy.mean():.3f} , F1 : {macro_f1.mean():.3f}")
   print(f"{name} --> Recall {recall_score.mean():.3f} , Precision : {precision_score.mean():.3f}")
   print("\n"*2)

```

```python
from sklearn.metrics import accuracy_score,f1_score,recall_score,precision_score,confusion_matrix
for name,model in models.items():
   pipe=Pipeline(steps=[
       ('preprocessor',preprocessor),('classifier',model)
   ])
   pipe.fit(X_train,y_train)
   y_pred=pipe.predict(X_test)
   print(f"*"*25,name,"*"*25)
   print(f"Accuracy            --> {accuracy_score(y_test,y_pred):.3f}")
   print(f"F1 Score            --> {f1_score(y_test,y_pred):.3f}")
   print(f"Recall Score Malign --> {recall_score(y_test,y_pred,pos_label=1):.3f}")
   print(f"Recall Score benign --> {recall_score(y_test,y_pred,pos_label=0):.3f}")
   print(f"Precision           --> {precision_score(y_test,y_pred):.3f}")
   print("\n"*2)
   plt.title(f"{name} Confusion Matrix")
   sns.heatmap(confusion_matrix(y_test,y_pred),annot=True,cmap="Blues",fmt="d")
   plt.xlabel("Predicted")
   plt.ylabel("Actual")
   plt.show()
   print("\n"*3)
```

# **17. HYPERPARAMETER TUNING**

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer
```

```python
best_pipeline=Pipeline(steps=[
       ('preprocessor',preprocessor),('classifier',LogisticRegression(max_iter=5000))
   ])
malign_recall=make_scorer(recall_score,pos_label=1)

param_grid={
    'classifier__C':[0.001,0.01,0.1,1,10,100],
    'classifier__class_weight':['balanced',None],
    'classifier__solver':['liblinear','saga','lgfgs','sag','newton-cg'],
    'classifier__penalty':['l1','l2']
}
grid_search=GridSearchCV(estimator=best_pipeline,param_grid=param_grid,cv=5,scoring=malign_recall,n_jobs=-1)
grid_search.fit(X_train,y_train)

```

```python
print("best estimator :",grid_search.best_params_)
print("best score :",grid_search.best_score_)
```

```python
plt.title('Best Estimator Model ')
sns.heatmap(confusion_matrix(y_test,grid_search.best_estimator_.predict(X_test)),annot=True,cmap="Blues",fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
y_pred=grid_search.best_estimator_.predict(X_test)
recall=recall_score(y_test,y_pred)
precision=precision_score(y_test,y_pred,average='macro')
f1=f1_score(y_test,y_pred)
accuracy=accuracy_score(y_test,y_pred)
print(f"Recall : {recall:.3f}")
print(f"Precision : {precision:.3f}")
print(f"F1 : {f1:.3f}")
print(f"Accuracy : {accuracy:.3f}")
```

# **18. RESULT AFTER HYPERPARAMETER TUNING**
Hyperparamter tuning is performed using GridSearchCV with 5-fold cross-validation, optimizing malignant-class recall. However, the tuned model did not outperform the baseline Logistic Regression Model on the held-out test set.

# **19. CONCLUSION**
Baseline Regression model was retained as the final model.
